```python
import random

def count_conflicts(board, n):
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


def print_board(board, n):
    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def get_neighbors(board, n):
    neighbors = []

    for col in range(n):
        for row in range(n):
            if board[col] != row:
                new_board = board.copy()
                new_board[col] = row
                neighbors.append(new_board)

    return neighbors


def hill_climbing(n):
    if n < 1:
        print("Invalid Input")
        return

    # Generate initial random board
    board = [random.randint(0, n - 1) for _ in range(n)]

    print("Initial Board:")
    print_board(board, n)

    conflicts = count_conflicts(board, n)
    print("\nNumber of attacking pairs:", conflicts)

    print("\nApplying Hill Climbing...")

    while conflicts != 0:
        neighbors = get_neighbors(board, n)

        best_board = min(
            neighbors,
            key=lambda b: count_conflicts(b, n)
        )

        best_conflicts = count_conflicts(best_board, n)

        if best_conflicts < conflicts:
            board = best_board
            conflicts = best_conflicts
        else:
            # Random restart
            board = [random.randint(0, n - 1) for _ in range(n)]
            conflicts = count_conflicts(board, n)

    print("\nSolution Found!")

    print("\nFinal Board:")
    print_board(board, n)

    print("\nNumber of attacking pairs:", conflicts)


n = int(input("Enter the value of N: "))
hill_climbing(n)



```
