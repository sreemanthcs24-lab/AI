```python

import random
import math

N = 8

def calculate_conflicts(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


def simulated_annealing():
    max_restarts = 100

    for restart in range(max_restarts):

        board = [random.randint(0, N - 1) for _ in range(N)]
        temperature = 2.0
        cooling_rate = 0.995

        for step in range(5000):
            current_cost = calculate_conflicts(board)

            if current_cost == 0:
                return board, restart + 1, step

            
            col = random.randint(0, N - 1)
            old_row = board[col]
            new_row = random.randint(0, N - 1)

            board[col] = new_row
            new_cost = calculate_conflicts(board)
            delta = new_cost - current_cost

        
            if (delta <= 0 or
                random.random() < math.exp(-delta / temperature)):
                pass
            else:
                board[col] = old_row

            
            temperature *= cooling_rate

        

    return None, max_restarts, 5000


def print_board(board):
    print("\nChessboard solution:")
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q", end="")
            else:
                print(" .", end="")
        print()



solution, restarts, steps = simulated_annealing()

if solution is not None:
    print("Solution found!")
    print("Queen positions (row per column):", solution)
    print("Conflicts:", calculate_conflicts(solution))
    print("Restarts:", restarts)
    print("Steps in final restart:", steps)
    print_board(solution)
else:
    print("No solution found. Run the program again.")
```
