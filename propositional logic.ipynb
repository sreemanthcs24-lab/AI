```python
from itertools import product

# Implication function
def implies(a, b):
    return (not a) or b

# Generate all possible truth combinations
print("TRUTH TABLE FOR PROPOSITIONAL LOGIC\n")

print("P\tQ\tR\tQ->P\tP->~Q\tQ|R\tKB")

models = []

for P, Q, R in product([True, False], repeat=3):

    # Knowledge Base sentences
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    # Complete Knowledge Base
    KB = s1 and s2 and s3

    print(
        P, Q, R,
        s1, s2, s3, KB,
        sep="\t"
    )

    if KB:
        models.append((P, Q, R))

# Display KB models
print("\nMODELS WHERE KB IS TRUE:")
for model in models:
    print(model)

# Entailment checking function
def entails(query):
    for P, Q, R in models:
        if not query(P, Q, R):
            return False
    return True

# Check the given queries
queries = {
    "KB entails R": lambda P, Q, R: R,
    "KB entails R -> P": lambda P, Q, R: implies(R, P),
    "KB entails Q -> R": lambda P, Q, R: implies(Q, R)
}

print("\nENTAILMENT RESULTS:")

for name, query in queries.items():
    if entails(query):
        print(name, ": TRUE")
    else:
        print(name, ": FALSE")

```
