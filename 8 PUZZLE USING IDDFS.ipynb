```python
import copy

GOAL_STATE = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


def find_blank(state):
    """Find the coordinates (row, col) of the blank space (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c


def get_successors(state):
    """Generate all valid next states by moving the blank space."""
    successors = []

    r, c = find_blank(state)

    moves = [
        (-1, 0, 'Up'),
        (1, 0, 'Down'),
        (0, -1, 'Left'),
        (0, 1, 'Right')
    ]

    for dr, dc, move_name in moves:

        nr = r + dr
        nc = c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_state = copy.deepcopy(state)

            new_state[r][c], new_state[nr][nc] = \
                new_state[nr][nc], new_state[r][c]

            successors.append((new_state, move_name))

    return successors


def depth_limited_dfs(state, goal, depth, path):
    """DFS with a depth limit."""

    if state == goal:
        return path

    if depth <= 0:
        return None

    for next_state, move in get_successors(state):

        if next_state not in [p[0] for p in path]:

            result = depth_limited_dfs(
                next_state,
                goal,
                depth - 1,
                path + [(next_state, move)]
            )

            if result is not None:
                return result

    return None


def iddfs(start_state, goal):
    """Iteratively increase the depth limit."""

    depth = 0

    while True:

        initial_path = [(start_state, "Initial")]

        result = depth_limited_dfs(
            start_state,
            goal,
            depth,
            initial_path
        )

        if result is not None:
            return result, depth

        depth += 1



if __name__ == "__main__":

    # Initial state
    initial_state = [
        [1, 2, 3],
        [4, 5, 0],
        [7, 8, 6]
    ]

    print("Searching for a solution using IDDFS...")

    solution_path, total_depth = iddfs(
        initial_state,
        GOAL_STATE
    )

    print(f"\nSolution found at Depth: {total_depth}")

    for index, (state, move) in enumerate(solution_path):

        print(f"\nStep {index}: {move}")

        for row in state:
            print(row)
```
