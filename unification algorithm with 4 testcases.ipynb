```python
def is_variable(x):
    return isinstance(x, str) and x.islower()


def unify(x, y, substitution):

    # If both are equal
    if x == y:
        return substitution

    # If x is a variable
    if is_variable(x):
        substitution[x] = y
        return substitution

    # If y is a variable
    if is_variable(y):
        substitution[y] = x
        return substitution

    # If both are different constants
    return None


def unification(expr1, expr2):

    # Check predicate names
    if expr1[0] != expr2[0]:
        return "FAIL"

    # Check number of arguments
    if len(expr1[1]) != len(expr2[1]):
        return "FAIL"

    substitution = {}

    # Compare corresponding arguments
    for x, y in zip(expr1[1], expr2[1]):

        result = unify(x, y, substitution)

        if result is None:
            return "FAIL"

    return substitution


# Test Case 1
E1 = ("Eats", ["x", "Apple"])
E2 = ("Eats", ["Riya", "y"])

# Test Case 2
E3 = ("Likes", ["x", "Mango"])
E4 = ("Likes", ["Rahul", "Mango"])

# Test Case 3
E5 = ("Eats", ["Apple", "Mango"])
E6 = ("Eats", ["Banana", "Mango"])

# Test Case 4
E7 = ("Eats", ["x", "Apple"])
E8 = ("Likes", ["Riya", "Apple"])


# Execute all 4 test cases
test_cases = [
    (E1, E2),
    (E3, E4),
    (E5, E6),
    (E7, E8)
]

for i, (expr1, expr2) in enumerate(test_cases, 1):

    print("\nTest Case", i)
    print("Expression 1:", expr1)
    print("Expression 2:", expr2)

    result = unification(expr1, expr2)

    if result == "FAIL":
        print("Unification Failed")
    else:
        print("Unification Successful")
        print("Substitution:", result)


```
